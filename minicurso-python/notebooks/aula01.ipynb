## Decisões: `if`, `elif` e `else`

```python
a, b, c = 1, -5, 6
delta = b**2 - 4*a*c
if delta > 0:
    x1 = (-b + delta**0.5) / (2*a)
    x2 = (-b - delta**0.5) / (2*a)
    print(f"Duas raízes reais: {x1} e {x2}")
elif delta == 0:
    print("Uma raiz real dupla:", -b / (2*a))
else:
    print("Não há raízes reais (delta < 0)")
print("delta =", delta)
```

```python
n = 15
if n % 2 == 0:
    print(n, "é par")
else:
    print(n, "é ímpar")

if n % 3 == 0 and n % 5 == 0:
    print(n, "é múltiplo de 3 e de 5")
```

## Repetições com `for`

```python
print(list(range(5)))
print(list(range(2, 11, 3)))
print(list(range(10, 0, -2)))
```

```python
for i in range(1, 6):
    print(i, i**2, i**3)
```

```python
n = 10
soma = 0                  # acumulador
for k in range(1, n + 1):
    soma = soma + k**2    # ou: soma += k**2
print("soma =", soma, "  fórmula =", n*(n + 1)*(2*n + 1)//6)
```

```python
n = 6
fat = 1
for k in range(2, n + 1):
    fat *= k
print(f"{n}! = {fat}")
```

```python
for letra in "MAT":
    print(letra, end=" ")
```

```python
# repetições encaixadas: tabuada de 1 a 3
for i in range(1, 4):
    for j in range(1, 6):
        print(f"{i*j:3d}", end="")
    print()   # pula linha ao fim de cada i
```

## Repetições com `while`

```python
soma = 0
k = 0
while soma <= 100:
    k += 1
    soma += k
print("k =", k, "  soma =", soma)
```

```python
a, b = 252, 198
while b != 0:
    print(f"mdc({a}, {b})")
    a, b = b, a % b
print("mdc =", a)
```

```python
# primeiro múltiplo de 7 maior que 50
n = 51
while True:          # repetição "infinita"...
    if n % 7 == 0:
        break        # ...interrompida pelo break
    n += 1
print(n)

for i in range(1, 11):
    if i % 3 == 0:
        continue     # pula os múltiplos de 3
    print(i, end=" ")
```
